Once `pyodbc` is installed, you can create a SQLAlchemy engine to connect to your SQL Server instance. Remember to replace the placeholder values with your actual SQL Server credentials. It's highly recommended to store sensitive information like passwords using Colab's 'Secrets' feature.

```python
from sqlalchemy import create_engine
import urllib

# --- Replace these with your actual SQL Server credentials ---
SQL_SERVER_HOST = "your_sql_server_host" # e.g., 'yourserver.database.windows.net' or 'localhost'
SQL_SERVER_PORT = "1433"                # Default SQL Server port
SQL_SERVER_USER = "your_sql_server_username"
SQL_SERVER_PASSWORD = "your_sql_server_password" # Use Colab Secrets for this!
SQL_SERVER_DB_NAME = "your_sql_server_database" # The name of your database
# ---------------------------------------------------------------

# URL-encode the password if it contains special characters
# If using Colab secrets, the password might already be safe, but this is good practice.
quoted_password = urllib.parse.quote_plus(SQL_SERVER_PASSWORD)

# Construct the connection string for SQL Server
# Using 'mssql+pyodbc' as the dialect. You might need to specify a driver.
# Common drivers include '{ODBC Driver 17 for SQL Server}' or '{SQL Server}'
# Check your SQL Server configuration for the correct driver name.
connection_string = (
    f'mssql+pyodbc://{SQL_SERVER_USER}:{quoted_password}@'
    f'{SQL_SERVER_HOST}:{SQL_SERVER_PORT}/{SQL_SERVER_DB_NAME}'
    '?driver={ODBC Driver 17 for SQL Server}'
)

try:
    # Create a SQLAlchemy engine
    sql_server_engine = create_engine(connection_string)

    # Test the connection
    with sql_server_engine.connect() as connection:
        print("Successfully connected to the SQL Server database!")

    # Example: Write DataFrame to a table in the SQL Server database
    # 'customer_data_sqlserver' will be the name of your table
    # Make sure 'df' DataFrame is available in your environment
    df.to_sql('customer_data_sqlserver', con=sql_server_engine, if_exists='replace', index=False)
    print(f"DataFrame successfully written to table 'customer_data_sqlserver' in database '{SQL_SERVER_DB_NAME}'.")

except Exception as e:
    print(f"Error connecting to or writing to SQL Server: {e}")
    print("Please ensure your host, port, username, password, and database name are correct.")
    print("Also, verify that the SQL Server is accessible from Colab and that the ODBC driver name is correct.")
```
